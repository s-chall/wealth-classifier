```python
import pandas as pd

YEAR = 2025
COUNTRY = "Poland"

BITCOIN_URL = "https://ccaf.io/cbeci/api/v1.2.0/download/data?price=0.05"
COUNTRY_URL = "https://raw.githubusercontent.com/owid/energy-data/master/owid-energy-data.csv"
```

```python
bitcoin_data = pd.read_csv(BITCOIN_URL, skiprows=1)
bitcoin_data["year"] = pd.to_datetime(bitcoin_data["Date and Time"]).dt.year

bitcoin_year = bitcoin_data[bitcoin_data["year"] == YEAR]
bitcoin_number = round(bitcoin_year["annualised consumption GUESS, TWh"].mean())

bitcoin_number
```

```python
country_data = pd.read_csv(COUNTRY_URL, usecols=["country", "year", "electricity_demand"])

country_row = country_data[(country_data["country"] == COUNTRY) & (country_data["year"] == YEAR)]
country_number = round(country_row["electricity_demand"].iloc[0])

country_number
```
